## 표준국어대사전 arrow 리소스 만들기

### 표준국어대사전 살펴보기

- `전체 내려받기_표준국어대사전_xls_20260304.zip`

```python
from glob import glob
from natsort import natsorted
import pandas as pd

# 엑셀 파일 로드 및 병합
xls_files = natsorted(glob("*.xls"))
if not xls_files:
    raise FileNotFoundError("코랩 환경에 표준국어대사전 엑셀 파일이 없습니다.")

print(f"총 {len(xls_files)}개의 엑셀 파일을 병합합니다...")
df_list = [pd.read_excel(xls) for xls in xls_files]
df = pd.concat(df_list, ignore_index=True)
print(f"전체 등재 데이터 건수: {len(df)}건\n")

df.sample(10)
```

```python
df["구성 단위"].drop_duplicates()
```

```python
# 단어 단위만 추출 (구, 관용구, 속담 제외)
word_df = df[df["구성 단위"] == "단어"].reset_index(drop=True)
print(f"추출된 단어 수: {len(word_df)}건\n")

word_df.sample(10)
```

```python
import re

# 한글, 영문, 숫자, 한자, 공백을 제외한 모든 문자를 찾는 정규표현식
symbol_regex = re.compile(r'[^ㄱ-ㅎㅏ-ㅣ가-힣a-zA-Z0-9一-龥\s]')

def extract_symbols(df, column_name):
    symbols = set()

    # NaN 결측치를 빈 문자열로 처리하며 전수조사
    for text in df[column_name].fillna(''):
        found = symbol_regex.findall(str(text))
        symbols.update(found)

    return sorted(list(symbols))

# 어휘 컬럼 기호 조사
vocab_symbols = extract_symbols(word_df, "어휘")
print(f"\n[어휘] 컬럼 내 특수기호 (총 {len(vocab_symbols)}개):")
print(vocab_symbols)

unexpected_symbols = [s for s in vocab_symbols if s not in ['(', ')', '-', '^', 'ㆍ']]
if unexpected_symbols:
    print(f"\n3년 전에는 없었던 [어휘] 기호 발견: {unexpected_symbols}")
    sample_df = word_df[word_df["어휘"].str.contains('|'.join([re.escape(s) for s in unexpected_symbols]), na=False)]
    display(sample_df[["어휘", "발음"]].sample(10))
del unexpected_symbols

# 발음 컬럼 기호 조사
pronun_symbols = extract_symbols(word_df, "발음")
print(f"\n[발음] 컬럼 내 특수기호 (총 {len(pronun_symbols)}개):")
print(pronun_symbols)

unexpected_symbols = [s for s in pronun_symbols if s not in ['/', '[', ']', 'ː']]
if unexpected_symbols:
    print(f"\n3년 전에는 없었던 [어휘] 기호 발견: {unexpected_symbols}")
    sample_df = word_df[word_df["어휘"].str.contains('|'.join([re.escape(s) for s in unexpected_symbols]), na=False)]
    display(sample_df[["어휘", "발음"]].sample(3)) # when >=4, ValueError: Cannot take a larger sample than population when 'replace=False'
del unexpected_symbols
```

### DAT 생성

```python
# 데이터 파싱용 헬퍼 함수 정의
def parse_pronunciation(p):
    p = str(p)

    # 판다스(Pandas)는 엑셀의 빈 칸을 float 타입의 NaN으로 읽어 들입니다.
    # 이를 str()로 변환하면 "nan"이라는 문자열이 되므로,
    # 발음 기호가 아예 기재되지 않은 단어들은 빈 문자열('')로 처리하여 후속 로직의 에러를 방지합니다.
    if p == "nan":
        return ''

    # 사전의 발음 표기는 보통 '[가ː]' 형태를 띱니다.
    # 형태소 경계와 음운 변동만 추출하는 음운변동(G2P) 목적상,
    # 1) 감싸고 있는 대괄호('[' 및 ']')를 제거하고
    # 2) 음의 장단을 나타내는 장음 기호('ː')를 제거하며
    # 3) 타이피스트의 오타인 일반 콜론(':')도 함께 제거하고
    # 4) 혹시 모를 오타성 공백(' ')을 모두 날려 순수한 한글 발음 텍스트만 남깁니다.
    p = p.replace('[', '').replace(']', '').replace('ː', '').replace(':', '').replace(' ', '')

    # '맛있다'처럼 복수 표준 발음이 인정되는 경우 사전에는 '[마디따/마시따]'로 등재됩니다.
    # 이 경우 슬래시('/')를 기준으로 문자열을 쪼개어 리스트를 만든 뒤,
    # [0]번 인덱스를 가져와 무조건 첫 번째 대표 발음(예: "마디따")만 취합니다.
    return p.split('/')[0]


def parse_compound(c):
    c = str(c)

    # '-가(12)'처럼 접사나 어미로 분류되어 표제어 맨 앞에 하이픈('-')이 붙어있는 경우가 있습니다.
    # 사전 검색 키워드(Key)로 사용하기 위해 맨 앞의 하이픈을 슬라이싱(c[1:])으로 잘라냅니다.
    if c.startswith('-'):
        c = c[1:]

    # 국어사전은 복합어의 내부 경계를 삿갓('^')이나 가운뎃점('ㆍ', '·')으로 표시합니다.
    # 꼼수로 쓰인 옛한글 아래아(ㆍ)와 진짜 가운뎃점(·)을 모두 단일 하이픈('-')으로 통일하여,
    # 추후 유음화(제20항) 엔진이 형태소 경계를 일관되게 인식할 수 있도록 정규화합니다.
    c = c.replace('^', '-').replace('ㆍ', '-').replace('·', '-')

    # 동음이의어 구분을 위한 번호(예: '가가대소(01)')를 제거하는 핵심 트릭입니다.
    # 여는 괄호('(')를 기준으로 split 하면 ["가가대소", "01)"] 형태의 리스트가 반환됩니다.
    # 여기서 [0]번 인덱스를 가져오면, 닫는 괄호(')')를 별도로 replace 처리할 필요 없이
    # "가가대소"라는 순수 표제어만 아주 깔끔하고 빠르게 분리해 낼 수 있습니다.
    return c.split('(')[0]
```

```python
from tqdm.auto import tqdm

# DAT 생성을 위한 딕셔너리 조립
nonkor = re.compile("[^ㄱ-ㅎㅏ-ㅣ가-힣]")
data_for_trie = {}

for _, row in tqdm(word_df.iterrows(), total=len(word_df), desc="DAT 사전 파싱 중"):
    raw_word = str(row["어휘"])

    # 유니코드 PUA 영역(옛한글 특수문자 \ue000~\uf8ff 등)이 포함된 옛날 단어는 통째로 버림
    if re.search(r'[\ue000-\uf8ff]', raw_word):
        continue

    clean_key = nonkor.sub('', raw_word)

    # 중복 표제어의 경우 첫 번째 기준을 채택
    if clean_key not in data_for_trie and clean_key != '':
        # 한자어 여부 판별
        is_hanja_val = '1' if (row["고유어 여부"] == "한자어" or "한자" in str(row["원어·어종"])) else '0'
        pronun = parse_pronunciation(row["발음"])
        comp = parse_compound(raw_word)

        data_for_trie[clean_key] = {
            "is_hanja": is_hanja_val,
            "pronunciation": pronun,
            "compound_structure": comp
        }
```

```python
from pecab._datrie import DoubleArrayTrie
import pyarrow as pa

print("Double-Array Trie 구조로 컴파일 중...")
# 딕셔너리를 DAT 객체로 변환
trie = DoubleArrayTrie(data_for_trie)

print("Arrow 바이너리 파일로 직렬화 중...")
# Words 테이블 변환
new_words = pd.DataFrame.from_records(list(trie._value))
new_words = pa.Table.from_pandas(new_words)

# Arrays 테이블(Base & Check) 변환
new_arrays = {"base": trie._base, "check": trie._check}
new_arrays = pd.DataFrame.from_dict(new_arrays)
new_arrays = pa.Table.from_pandas(new_arrays)

words_out = "stdict_words.arrow"
arrays_out = "stdict_arrays.arrow"

with pa.OSFile(words_out, "wb") as sink:
    with pa.RecordBatchFileWriter(sink, new_words.schema) as writer:
        writer.write_table(new_words)

with pa.OSFile(arrays_out, "wb") as sink:
    with pa.RecordBatchFileWriter(sink, new_arrays.schema) as writer:
        writer.write_table(new_arrays)

print(f"'stdict_words.arrow' 및 'stdict_arrays.arrow'가 생성되었습니다.")
```
